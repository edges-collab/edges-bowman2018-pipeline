```python
from pygsdata import GSData
from pygsdata.select import select_freqs
from astropy import units as un

data = GSData.from_file("output-notebooks/2016-260.pre-cal.gsh5")

print("Number of operations applied to this data: ", len(data.history))
print()

print("Last operation applied to the data:\n-------------------------")
print(data.history[-1])

print(f"Original: Nfreqs={data.nfreqs}, Ntimes={data.ntimes}")

down_selected = select_freqs(data, freq_range=(60*un.MHz, 70*un.MHz))
print("New number of frequencies: ", down_selected.nfreqs)
```
